```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
```

```python
train_modelo = train.copy()
test_modelo = test.copy()
banco_modelo = cbanco.copy()
```

```python
train_modelo["Fecha_Solicitud"] = train_modelo["Fecha_Solicitud"].astype(str)
test_modelo["Fecha_Solicitud"] = test_modelo["Fecha_Solicitud"].astype(str)
banco_modelo["Mes_Referencia"] = banco_modelo["Mes_Referencia"].astype(str)
```

```python
def agregarHistorialBancario(solicitudes, historial):
    datos = solicitudes[
        ["ID_Solicitud", "ID_Usuario", "Fecha_Solicitud"]
    ].join(
        historial.set_index("ID_Usuario"),
        on="ID_Usuario",
        how="left"
    )
    datos = datos[
        datos["Mes_Referencia"] < datos["Fecha_Solicitud"]
    ]

    resumen = []
    for id_solicitud in solicitudes["ID_Solicitud"].unique():
        grupo = datos[datos["ID_Solicitud"] == id_solicitud]

        saldo_promedio = grupo["Saldo_Promedio_Mensual"].mean()
        saldo_minimo = grupo["Saldo_Promedio_Mensual"].min()
        saldo_maximo = grupo["Saldo_Promedio_Mensual"].max()

        atraso_promedio = grupo["Dias_Atraso_Otras_Deudas"].mean()
        atraso_maximo = grupo["Dias_Atraso_Otras_Deudas"].max()

        meses_historial = grupo["Mes_Referencia"].nunique()

        resumen.append({
            "ID_Solicitud": id_solicitud,
            "saldo_promedio": saldo_promedio,
            "saldo_minimo": saldo_minimo,
            "saldo_maximo": saldo_maximo,
            "atraso_promedio": atraso_promedio,
            "atraso_maximo": atraso_maximo,
            "meses_historial": meses_historial
        })
    resumen = pd.DataFrame(resumen)
    resumen = resumen.set_index("ID_Solicitud")
    
    resultado = solicitudes.join(
        resumen,
        on="ID_Solicitud",
        how="left"
    )
    return resultado
train_preparado = agregarHistorialBancario(train_modelo, banco_modelo)
test_preparado = agregarHistorialBancario(test_modelo, banco_modelo)
```

```python
entrenamiento = train_preparado[
    train_preparado["Fecha_Solicitud"] < "2025-11"
].copy()

validacion = train_preparado[
    train_preparado["Fecha_Solicitud"] >= "2025-11"
].copy()
```

```python
print("Solicitudes entrenar:", len(entrenamiento))
print("Solicitudes validar:", len(validacion))
```

```python
print("\nMeses entrenamiento:")
print(entrenamiento[["Fecha_Solicitud"]].drop_duplicates().sort_values("Fecha_Solicitud"))

print("\nMeses validación:")
print(validacion[["Fecha_Solicitud"]].drop_duplicates().sort_values("Fecha_Solicitud"))
```

```python
print("\nProporción pagos entrenamiento:")
print(entrenamiento["Estado_Final"].mean())

print("\nProporción pagos validación:")
print(validacion["Estado_Final"].mean())
```

```python
columnas_numericas = [
    "Monto_Solicitado",
    "Tasa_Interes",
    "Ingreso_Mensual",
    "saldo_promedio",
    "saldo_minimo",
    "saldo_maximo",
    "atraso_promedio",
    "atraso_maximo",
    "meses_historial"
]
columnas_categoricas = ["Motivo_Prestamo"]
columnas_modelo = columnas_numericas + columnas_categoricas

X_entrenamiento = entrenamiento[columnas_modelo]
y_entrenamiento = entrenamiento["Estado_Final"]

X_validacion = validacion[columnas_modelo]
y_validacion = validacion["Estado_Final"]

print("Variables entrada:", columnas_modelo)
print("X_entrenamiento:", X_entrenamiento.shape)
print("X_validacion:", X_validacion.shape)
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
preprocessor = ColumnTransformer([
    ("numericas", numeric_pipeline, columnas_numericas),
    ("categoricas", categorical_pipeline, columnas_categoricas)
])
modelo = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ))
])

modelo.fit(X_entrenamiento, y_entrenamiento)

print("Modelo entrenado correctamente")
```

```python
from sklearn.metrics import roc_auc_score, accuracy_score, classification_report
probabilidades_validacion = modelo.predict_proba(
    X_validacion
)[:, 1]

predicciones_validacion = (
    probabilidades_validacion >= 0.5
).astype(int)

auc = roc_auc_score(
    y_validacion,
    probabilidades_validacion
)

accuracy = accuracy_score(
    y_validacion,
    predicciones_validacion
)

print("AUC:", auc)
print("Accuracy:",accuracy)

print("\nReporte de clasificación:")
print(classification_report(
    y_validacion,
    predicciones_validacion,
    target_names=["Incumple (0)", "Paga (1)"]
))
```

```python
resultados = validacion[
    ["ID_Solicitud", "Monto_Solicitado", "Tasa_Interes", "Estado_Final"]
].copy()

resultados["Probabilidad_Pago"] = probabilidades_validacion

resultados["Ganancia_Esperada"] = (
    resultados["Probabilidad_Pago"]
    * resultados["Monto_Solicitado"]
    * resultados["Tasa_Interes"]
    - (1 - resultados["Probabilidad_Pago"])
    * resultados["Monto_Solicitado"]
)

resultados["Rendimiento"] = (
    resultados["Ganancia_Esperada"]
    / resultados["Monto_Solicitado"]
)

print("Resultados:")
display(resultados.head())
```

```python
candidatos = resultados[
    resultados["Ganancia_Esperada"] > 0
].sort_values("Rendimiento", ascending=False)

presupuesto = 2_000_000
capital_usado = 0
indices_aprobados = []

for indice in candidatos.index:
    monto = candidatos.loc[indice, "Monto_Solicitado"]

    if capital_usado + monto <= presupuesto:
        indices_aprobados.append(indice)
        capital_usado += monto

aprobados_validacion = resultados.loc[indices_aprobados].copy()

print("Cantidad de préstamos aprobados:", len(aprobados_validacion))
print("Capital utilizado: $", capital_usado)
print("Presupuesto restante: $", presupuesto - capital_usado)
```

```python
X = train_preparado[columnas_modelo]
y = train_preparado["Estado_Final"]

modelo_final = Pipeline([
    ("preprocessor", preprocessor),
    ("modelo_final", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ))
])

modelo_final.fit(X, y)

print("Modelo entrenado.")
print("Solicitudes utilizadas:", len(X))
```

```python
X_test = test_preparado[columnas_modelo]
probabilidades_test = modelo_final.predict_proba(X_test)[:, 1]

print("Cantidad de predicciones:", len(probabilidades_test))
print("Primeras 10 probabilidades:")
print(probabilidades_test[:10])
```

```python
seleccion = test_preparado[
    ["ID_Solicitud", "Monto_Solicitado", "Tasa_Interes"]
].copy()

seleccion["Probabilidad_Pago"] = probabilidades_test

seleccion["Ganancia_Esperada"] = (
    seleccion["Probabilidad_Pago"]
    * seleccion["Monto_Solicitado"]
    * seleccion["Tasa_Interes"]
    - (1 - seleccion["Probabilidad_Pago"])
    * seleccion["Monto_Solicitado"]
)

seleccion["Rendimiento"] = (
    seleccion["Ganancia_Esperada"]
    / seleccion["Monto_Solicitado"]
)

candidatos = seleccion[
    seleccion["Ganancia_Esperada"] > 0
].sort_values("Rendimiento", ascending=False)

presupuesto = 2_000_000
capital_usado = 0
aprobados = []

for i in range(len(candidatos)):
    monto = candidatos["Monto_Solicitado"].iloc[i]
    id_solicitud = candidatos["ID_Solicitud"].iloc[i]

    if capital_usado + monto <= presupuesto:
        aprobados.append(id_solicitud)
        capital_usado += monto

print("Préstamos aprobados:", len(aprobados))
print("Capital utilizado: $", capital_usado)
print("Presupuesto restante: $", presupuesto - capital_usado)
```

```python
submission_logistica = pd.DataFrame({
    "ID_Solicitud": test_preparado["ID_Solicitud"],
    "Decision_Aprobacion": (
        test_preparado["ID_Solicitud"]
        .isin(set(aprobados))
        .astype(int)
    )
})

submission_logistica.to_csv(
    "PrediccionRL.csv",
    index=False
)

print("Archivo creado: PrediccionRL.csv")
```
